```python
import pandas as pd

# Load the Excel file
excel_file = "EMA BD Team 2025.xlsx"  # Replace with your Excel file path
df = pd.read_excel(excel_file)

# Convert to JSON
json_file = "output.json"  # Specify the desired JSON file name
df.to_json(json_file, orient="records", indent=4)

print(f"Excel file has been converted to JSON and saved as {json_file}")
```
